## WhatsApp

```python
import pandas as pd 

# Lendo o CSV
df = pd.read_csv("../Datasets/Correto_whatsapp_rotulado_revisado.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df['id'] = range(len(df))
df.head()
```

```python
print(df.isnull().sum())
```

```python
df.shape
```

```python
# === Amostragem balanceada ===
df_0 = df[df["preconceito"] == 0].sample(n=300, random_state=42)
df_1 = df[df["preconceito"] == 1].sample(n=300, random_state=42)

# Combina e embaralha
df_sample = pd.concat([df_0, df_1]).sample(frac=1, random_state=42).reset_index(drop=True)

# === Garante que o restante NÃO tenha os mesmos textos ===
df_restante = df[~df["text_content_anonymous"].isin(df_sample["text_content_anonymous"])]

# === Salva ===
df_sample.to_csv("./datasets/wpp/amostra_balanceada_com_rotulo.csv", index=False)
df_restante.to_csv("./datasets/wpp/dados_restantes.csv", index=False)

print("Amostra balanceada:", df_sample.shape)
print("Restante:", df_restante.shape)
```

```python
duplicadas = set(df_sample["text_content_anonymous"]) & set(df_restante["text_content_anonymous"])
print("Mensagens duplicadas:", len(duplicadas))
```

```python
df_sample[["text_content_anonymous", "id"]].to_csv(
    "./datasets/wpp/amostra_balanceada_sem_rotulo.csv",
    index=False
)
```

```python
df_restante[["text_content_anonymous", "id"]].to_csv(
    "./datasets/wpp/df_restantes_sem_rotulos.csv",
    index=False
)
```
