```python
import pandas as pd
import math
```

## WhatsApp

```python

# Caminho do arquivo original
CSV_PATH = "./datasets/wpp/amostra_balanceada_sem_rotulo.csv"
df = pd.read_csv(CSV_PATH)

# Número de linhas por arquivo
LINHAS_POR_ARQUIVO = 50

# Número total de arquivos
num_arquivos = math.ceil(len(df) / LINHAS_POR_ARQUIVO)
print(f"Serão criados {num_arquivos} arquivos.")

# Loop para criar cada arquivo
for i in range(num_arquivos):
    start = i * LINHAS_POR_ARQUIVO
    end = start + LINHAS_POR_ARQUIVO
    df_part = df.iloc[start:end]
    nome_arquivo = f"./datasets/wpp/parte_{i+1}_wpp.csv"
    df_part.to_csv(nome_arquivo, index=False, encoding="utf-8")
    print(f"Arquivo {nome_arquivo} criado com {len(df_part)} linhas.")
```
