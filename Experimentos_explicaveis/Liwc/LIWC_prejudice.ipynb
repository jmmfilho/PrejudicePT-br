```python
import pandas as pd
import re
import liwc
from scipy.stats import ttest_ind
```

```python

def tokenize(text):
    return re.findall(r'\w+', str(text).lower(), re.UNICODE)

parse, category_names = liwc.load_token_parser('../Datasets/LIWC2007_Portugues_win.dic')
```

## Wpp

```python

df = pd.read_csv('../Datasets/Correto_whatsapp_rotulado_revisado.csv', sep=",")

df['preconceito'] = df['preconceito'].astype(str)
df['preconceito'] = df['preconceito'].replace({'1': 'Prejudice.', '0': 'No Prejudice'})

# Cria as colunas
for cat in category_names:
    df[cat] = 0

# Conta categorias
for i, text in enumerate(df['text_content_anonymous']):
    tokens = tokenize(text)
    counts = dict.fromkeys(category_names, 0)
    for token in tokens:
        for category in parse(token):
            counts[category] += 1
    for cat in category_names:
        df.at[i, cat] = counts[cat]

# Normaliza por número de palavras
df['n_words'] = df['text_content_anonymous'].apply(lambda x: len(tokenize(x)))
for cat in category_names:
    df[cat] = df[cat] / df['n_words']

# Calcula médias e testes
results = []
for cat in category_names:
    mis_values = df.loc[df['preconceito'] == 'Prejudice.', cat]
    nonmis_values = df.loc[df['preconceito'] == 'No Prejudice', cat]

    mis_mean = mis_values.mean()
    nonmis_mean = nonmis_values.mean()
    diff = abs(mis_mean - nonmis_mean)
    t, p = ttest_ind(mis_values, nonmis_values, equal_var=False, nan_policy='omit')
    sig = "Yes" if p < 0.05 else "No"

    results.append({
        'Category': cat,
        'Prejudice.': mis_mean,
        'No Prejudice.': nonmis_mean,
        'Diff.': diff,
        'Prejudice Percent': mis_mean * 100,
        'No Prejudice Percent': nonmis_mean * 100,
        'Diff. Percent': diff * 100,
        'Sig. Diff.': sig
    })

results_df = pd.DataFrame(results)
results_df = results_df.sort_values('Diff.', ascending=False).reset_index(drop=True)
pd.set_option('display.float_format', '{:.3f}'.format)
print(results_df.head(20))
```

```python
significant = results_df[results_df['Sig. Diff.'] == 'Yes']
print("Categorias com diferença significativa (p < 0.05):")
print(significant[['Category', 'Diff.', 'Diff. Percent']])
```

```python
nonsignificant = results_df[results_df['Sig. Diff.'] == 'No']
print("\nCategorias sem diferença significativa:")
print(nonsignificant[['Category', 'Diff.', 'Diff. Percent']])
```

```python
print("\nResumo:")
print(f"Com diferença significativa: {len(significant)} categorias")
print(f"Sem diferença significativa: {len(nonsignificant)} categorias")
```
